```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv('https://raw.githubusercontent.com/campusx-official/fastapi-demo-api/refs/heads/main/insurance.csv')
data.head(10)
```

```python
for i in data.index: print(f"Index: {i}, age: {data['age'][i]}, weight: {data['weight'][i]}, height: {data['height'][i]}, income_lpa: {data['income_lpa'][i]}, smoker: {data['smoker'][i]}, city: {data['city'][i]}, occupation: {data['occupation'][i]}, insurance_premium_category: {data['insurance_premium_category'][i]}")
```

```python
# Feature Engineering: Creating a new feature 'bmi' (Body Mass Index) based on weight and height

df = data.copy()

bmi = lambda x: round(x['weight'] / (x['height']) ** 2, 2)

df['bmi'] = df.apply(bmi, axis=1)
```

```python
# Feature Engineering: Creating a new feature 'age_group' based on age

age_group = lambda x: 'young' if x['age'] < 25 else ('adult' if 25 < x['age'] < 45 else ('middle-aged' if 45 < x['age'] < 60 else 'senior'))

df['age_group'] = df.apply(age_group, axis=1)
```

```python
# Feature Engineering: Creating a new feature 'age_group' based on age

lifestyle_risk = lambda x: 'high' if x['smoker'] and x['bmi']  > 30 else ('medium' if x['smoker'] and x['bmi'] > 27 else 'low')

df['lifestyle_risk'] = df.apply(lifestyle_risk, axis=1)
```

```python
# Feature Engineering: Creating a new feature 'city_tier' based on city names

tier_1_cities = ["Mumbai", "Delhi", "Bangalore", "Chennai", "Kolkata", "Hyderabad", "Pune"]
tier_2_cities = [
    "Jaipur", "Chandigarh", "Indore", "Lucknow", "Patna", "Ranchi", "Visakhapatnam", "Coimbatore",
    "Bhopal", "Nagpur", "Vadodara", "Surat", "Rajkot", "Jodhpur", "Raipur", "Amritsar", "Varanasi",
    "Agra", "Dehradun", "Mysore", "Jabalpur", "Guwahati", "Thiruvananthapuram", "Ludhiana", "Nashik",
    "Allahabad", "Udaipur", "Aurangabad", "Hubli", "Belgaum", "Salem", "Vijayawada", "Tiruchirappalli",
    "Bhavnagar", "Gwalior", "Dhanbad", "Bareilly", "Aligarh", "Gaya", "Kozhikode", "Warangal",
    "Kolhapur", "Bilaspur", "Jalandhar", "Noida", "Guntur", "Asansol", "Siliguri"
]
     
def city_tier(city:str)->int:
    if city in tier_1_cities:
        return 1
    elif city in tier_2_cities:
        return 2
    else:
        return 3

df["city_tier"] = df["city"].apply(city_tier)
```

```python
# Define categorical and numeric features
categorical_features = ["age_group", "lifestyle_risk", "occupation", "city_tier"]
numeric_features = ["bmi", "income_lpa"]
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

# Create column transformer for OHE
preprocessor = ColumnTransformer(
    transformers=[
        ("cat", OneHotEncoder(), categorical_features),
        ("num", "passthrough", numeric_features)
    ]
)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier

# Create a pipeline with preprocessing and random forest classifier
pipeline = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("classifier", RandomForestClassifier(random_state=42))
])
```

```python
from sklearn.model_selection import train_test_split

# Split data and train model
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)
pipeline.fit(X_train, y_train)
```

```python
from sklearn.metrics import classification_report, accuracy_score

y_pred = pipeline.predict(X_test)
print("Accuracy:", accuracy_score(y_test, y_pred)*100,'%')
print("Classification Report:")
print(classification_report(y_test, y_pred))
```

```python
for i in range(len(y_test)):
    print(f"Index: {i}, Actual: {y_test.iloc[i]}, Predicted: {y_pred[i]}")
    print("Correct Prediction" if y_test.iloc[i] == y_pred[i] else "Incorrect Prediction")
```

```python
import pickle

model_path = "insurance_premium_model.pkl"

with open(model_path, "wb") as f:
    pickle.dump(pipeline, f)
```
